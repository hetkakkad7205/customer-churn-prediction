```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv("cleaned_custumer_churn_data.csv")

#NUMERICAL FEATURE DISTRIBUTIONS
numeric_cols = [
    "Age",
    "TenureMonths",
    "MonthlyCharges",
    "TotalCharges",
    "SupportCalls"
]

for col in numeric_cols:
    plt.figure(figsize=(8, 5))

    sns.histplot(
        data=data,
        x=col,
        kde=True
    )

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv("cleaned_custumer_churn_data.csv")

# NUMERICAL FEATURES BY CHURN
for col in numeric_cols:
    plt.figure(figsize=(7, 5))

    sns.boxplot(
        data=data,
        x="Churn",
        y=col,
        hue="Churn",
        legend=False
    )

    plt.title(f"{col} by Churn Status")
    plt.xlabel("Churn")
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("cleaned_custumer_churn_data.csv")

numeric_cols = [
    "Age",
    "TenureMonths",
    "MonthlyCharges",
    "TotalCharges",
    "SupportCalls"
]

data["ChurnNumeric"] = (
    data["Churn"].astype(str).str.strip().str.lower()
    .map({"yes": 1, "no": 0, "1": 1, "0": 0})
)

numeric_data = data[numeric_cols + ["ChurnNumeric"]]

correlation = numeric_data.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
